## Importing necessary libraries


```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

### Task: Train a Linear Regression Model Using a Single Variable

```python
# importing dataset from house_dataset.csv and saving it in df variable

df=pd.read_csv("house_dataset.csv")
# Displying no of rows and columns
df.shape
```

```python
# Printing fews records
df.sample(8)
```

```python
# Now create a Linear Regression model
model=LinearRegression()
model
```

```python
# fit the dependent and independent variable
model.fit(df[["Area (sqft)"]],df["Price (Lakhs)"])
```

```python
# Now do predict 
model.predict([[1500],[700]])
```

```python
model.coef_,model.intercept_
```

## Task: Train a Linear Regression Model Using a Multiple Variable

```python
# importing dataset from house_dataset.csv and saving it in df variable
df=pd.read_csv("house_price_dataset_1000.csv")

# Displying no of rows and columns
df.shape
```

```python
# Printing fews records
df.sample(8)
```

```python
# Now create a Linear Regression model
model=LinearRegression()
model
```

```python
# fit the dependent and independent variable
model.fit(df[["Area (sqft)","Bedrooms"]],df["Price (Lakhs)"])
```

```python
model.coef_,model.intercept_
```

```python
test=pd.DataFrame({
"Area (sqft)":[5000,800,1200],
"Bedrooms":[19,4,7]}
)
```

```python
model.predict(test)
```
